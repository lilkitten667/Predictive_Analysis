### Create the training and testing data

Put 70% of the data into a training set and the remaining 30% into a testing set. Create an X and y DataFrame with only the necessary variables.

### Create a plot of your model

Create a plot of your model to visualize results using the seaborn package.

```python
### YOUR CODE HERE ###
sns.regplot(x = 'Inflight entertainment', y = 'satisfaction_satisfied', 
            data = df_subset, logistic = True, ci = None)

plt.title('Logistic Regression: Inflight Entertainment vs Satisfaction')

plt.show()
```

```python
# 1. Create a clean range of possible ratings (1 to 5)
x_range = np.linspace(1, 5, 100).reshape(-1, 1)

# 2. Use your trained model (clf) to get the satisfaction probabilities
y_probabilities = clf.predict_proba(x_range)[:, 1]

# 3. Plot the raw data points as a faint scatter plot
sns.scatterplot(x='Inflight entertainment', y='satisfaction_satisfied', data=df_subset, alpha=0.05, color='gray')

# 4. Draw your model's actual learned logistic curve over it
plt.plot(x_range, y_probabilities, color='blue', linewidth=3, label='Model Fit')

plt.xlabel('Inflight entertainment Rating')
plt.ylabel('Probability of Being Satisfied')
plt.title('Visualizing the Logistic Regression Model')
plt.legend()
plt.show()
```

### Analyze the results

Print out the model's accuracy, precision, recall, and F1 score.

```python
### YOUR CODE HERE ###

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 1. Generate predictions using your test features
y_pred = clf.predict(X_test)

# 2. Calculate the metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

# 3. Print the results clearly
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1 Score:  {f1:.4f}")
```

### Produce a confusion matrix

Data professionals often like to know the types of errors made by an algorithm. To obtain this information, produce a confusion matrix.

```python
### YOUR CODE HERE ###
import sklearn.metrics as metrics
# Calculating the values for each quadrant in the Confusion Matrix

cm = metrics.confusion_matrix(y_test, y_pred, labels = clf.classes_)

# Creating the confusion matrix as a visuzlization

disp = metrics.ConfusionMatrixDisplay(confusion_matrix = cm, display_labels = clf.classes_)

# Displaying the Confusion Matrix

disp.plot(values_format = 'd')

plt.show()
```
